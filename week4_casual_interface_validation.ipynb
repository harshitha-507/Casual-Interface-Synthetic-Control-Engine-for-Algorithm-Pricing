```python
import numpy as np

def calculate_smd(data, variables):
    treated = data[data['treatment'] == 1]
    control = data[data['treatment'] == 0]

    results = {}

    for col in variables:
        mean_t = treated[col].mean()
        mean_c = control[col].mean()

        var_t = treated[col].var()
        var_c = control[col].var()

        pooled_sd = np.sqrt((var_t + var_c) / 2)

        results[col] = (
            (mean_t - mean_c) / pooled_sd
            if pooled_sd != 0 else 0
        )

    return pd.Series(results)

smd_before = calculate_smd(df, adjustment_variables)
smd_after = calculate_smd(matched_df, adjustment_variables)

balance_results = pd.DataFrame({
    'Before Matching': smd_before,
    'After Matching': smd_after
})

balance_results['Absolute SMD After'] = (
    balance_results['After Matching'].abs()
)

display(balance_results)
```

```python
balance_variables = [
    'demand_index',
    'inventory_level',
    'promotion_flag',
    'order_hour'
]

smd_results = []

for var in balance_variables:
    treated_mean = matched_treated[var].mean()
    control_mean = matched_control[var].mean()

    treated_std = matched_treated[var].std()
    control_std = matched_control[var].std()

    pooled_std = np.sqrt(
        (treated_std**2 + control_std**2) / 2
    )

    smd = (treated_mean - control_mean) / pooled_std

    smd_results.append({
        'Variable': var,
        'SMD': smd,
        'Absolute SMD': abs(smd)
    })

smd_table = pd.DataFrame(smd_results)

display(smd_table)
```

```python
from scipy import stats
import numpy as np

# Calculate paired revenue differences
pair_differences = (
    matched_treated['revenue'].to_numpy()
    - matched_control['revenue'].to_numpy()
)

matched_ate = np.mean(pair_differences)

# Preliminary paired standard error
standard_error = (
    np.std(pair_differences, ddof=1)
    / np.sqrt(len(pair_differences))
)

# 95% confidence interval
critical_value = stats.t.ppf(
    0.975, df=len(pair_differences) - 1
)

lower_ci = matched_ate - critical_value * standard_error
upper_ci = matched_ate + critical_value * standard_error

print("Matched ATE:", matched_ate)
print("Standard Error:", standard_error)
print("95% Confidence Interval:", lower_ci, "to", upper_ci)
print("Number of Matched Pairs:", len(pair_differences))
```

### Week 4 – Uncertainty and Limitations

**1. Matched ATE**
- Matched ATE: 0.9514 revenue units.
- Standard error: 3.7914.
- Preliminary 95% confidence interval: -6.4851 to 8.3878.
- The confidence interval includes zero, so the estimated effect is not statistically distinguishable from zero at the 5% level using this preliminary method.

**2. Matching Quality**
- Total matched observations: 3,578.
- Treated observations: 1,789.
- Matched control observations: 1,789.
- Unique matched controls: 1,631.
- Repeated control matches: 158.
- All four measured covariates had absolute SMD values below 0.1.

**3. Regional Findings**
- North Metro preliminary revenue difference: +7.0560.
- West Metro preliminary revenue difference: -6.2443.
- These are descriptive regional differences, not validated CATE estimates.

**4. Limitations**
- Only North Metro and West Metro received treatment.
- Control observations were reused during matching.
- The preliminary confidence interval does not account for dependence caused by reused controls.
- Only four measured covariates were checked for balance.
- The results do not establish a definitive causal impact across all regions.

**Conclusion:** Matching achieved good balance on the four measured covariates. However, the estimated overall revenue effect remains statistically inconclusive, and additional validation is required before making definitive causal claims.

```python
import statsmodels.api as sm
from scipy import stats
import numpy as np

pair_df = pd.DataFrame({
    'revenue_difference': pair_differences,
    'control_id': matched_control.index.to_numpy()
})

X_intercept = np.ones((len(pair_df), 1))

model = sm.OLS(
    pair_df['revenue_difference'],
    X_intercept
).fit(
    cov_type='cluster',
    cov_kwds={'groups': pair_df['control_id']}
)

cluster_se = np.sqrt(model.cov_params()[0, 0])

num_clusters = pair_df['control_id'].nunique()
critical_value = stats.t.ppf(0.975, num_clusters - 1)

lower = matched_ate - critical_value * cluster_se
upper = matched_ate + critical_value * cluster_se

print("Matched ATE:", matched_ate)
print("Cluster-adjusted SE:", cluster_se)
print("95% Confidence Interval:", lower, "to", upper)
print("Unique control clusters:", num_clusters)
```
